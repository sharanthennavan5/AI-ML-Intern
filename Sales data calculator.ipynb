```python
sales = [12500, 18200, 9800, 15600, 21300, 17400, 11900, 24500, 16800, 19200]

total_sales = sum(sales)
average_sales = total_sales / len(sales)
highest_sale = max(sales)
lowest_sale = min(sales)

print("Sales Data Analysis")
print("-" * 30)
print("Sales values:", sales)
print("Total sales:", total_sales)
print("Average sales:", round(average_sales, 2))
print("Highest sale:", highest_sale)
print("Lowest sale:", lowest_sale)

print("\nResult Summary")
print("-" * 30)
print(f"The total sales are {total_sales}.")
print(f"The average sale is {average_sales:.2f}.")
print(f"The highest sale is {highest_sale}.")
print(f"The lowest sale is {lowest_sale}.")
```

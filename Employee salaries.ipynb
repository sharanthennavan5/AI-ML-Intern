```python
import pandas as pd
import matplotlib.pyplot as plt

employees = [
    {"Name": "Arun", "Salary": 25000},
    {"Name": "Bala", "Salary": 32000},
    {"Name": "Charan", "Salary": 28000},
    {"Name": "Dinesh", "Salary": 45000},
    {"Name": "Ezhil", "Salary": 38000},
    {"Name": "Karthik", "Salary": 52000},
    {"Name": "Manoj", "Salary": 30000},
    {"Name": "Rahul", "Salary": 41000},
    {"Name": "Sanjay", "Salary": 35000},
    {"Name": "Vijay", "Salary": 60000}
]

salaries = [employee["Salary"] for employee in employees]

total_salary = sum(salaries)
average_salary = total_salary / len(salaries)
highest_salary = max(salaries)
lowest_salary = min(salaries)

print("=" * 45)
print("           EMPLOYEE SALARY ANALYSIS")
print("=" * 45)

for employee in employees:
    print(f"{employee['Name']:<15} ₹{employee['Salary']:,}")

print("\n" + "=" * 45)
print("             SALARY STATISTICS")
print("=" * 45)
print(f"Total Salary   : ₹{total_salary:,}")
print(f"Average Salary : ₹{average_salary:,.2f}")
print(f"Highest Salary : ₹{highest_salary:,}")
print(f"Lowest Salary  : ₹{lowest_salary:,}")

original_salaries = [25000, 32000, 28000, 45000, 38000, 52000, 30000, 41000, 35000, 60000]

print("\n" + "=" * 45)
print("          DATA COMPARISON")
print("=" * 45)
print("Original Data :", original_salaries)
print("Stored Data   :", salaries)
print("Data Match    :", original_salaries == salaries)
print("Total Match   :", sum(original_salaries) == total_salary)
print("Highest Match :", max(original_salaries) == highest_salary)
print("Lowest Match  :", min(original_salaries) == lowest_salary)

df = pd.DataFrame(employees)

print("\n" + "=" * 45)
print("             PANDAS ANALYSIS")
print("=" * 45)
display(df)

print(f"\nPandas Total Salary   : ₹{df['Salary'].sum():,}")
print(f"Pandas Average Salary : ₹{df['Salary'].mean():,.2f}")
print(f"Pandas Highest Salary : ₹{df['Salary'].max():,}")
print(f"Pandas Lowest Salary  : ₹{df['Salary'].min():,}")

highest_employee = df.loc[df["Salary"].idxmax()]
lowest_employee = df.loc[df["Salary"].idxmin()]

print("\n" + "=" * 45)
print("          EMPLOYEE DETAILS")
print("=" * 45)
print(f"Highest Paid Employee : {highest_employee['Name']} - ₹{highest_employee['Salary']:,}")
print(f"Lowest Paid Employee  : {lowest_employee['Name']} - ₹{lowest_employee['Salary']:,}")

plt.figure(figsize=(10, 5))
plt.bar(df["Name"], df["Salary"])
plt.xlabel("Employees")
plt.ylabel("Salary (₹)")
plt.title("Employee Salary Analysis")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("\n" + "=" * 45)
print("          INTERVIEW QUESTIONS")
print("=" * 45)

print("\n1. How do you calculate the average salary?")
print("Average Salary = Total Salary / Number of Employees")

print("\n2. What is the difference between total salary and average salary?")
print("Total salary is the sum of all salaries, while average salary is the total divided by the number of employees.")

print("\n3. How would you analyze salary data using Pandas?")
print("Pandas DataFrame can be used with sum(), mean(), max(), and min() functions to analyze salary data.")

print("\n" + "=" * 45)
print("             ANALYSIS COMPLETED")
print("=" * 45)
```
